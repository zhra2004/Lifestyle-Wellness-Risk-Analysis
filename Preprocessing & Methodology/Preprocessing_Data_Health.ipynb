```python
import numpy as np
import pandas as pd

# Contoh fungsi untuk membuat kolom Age Group
bins = [0, 30, 45, 60, 150]  # Batas atas usia
labels = ['Young Adults', 'Adults', 'Middle-Aged', 'Seniors']

df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=True)
```

```python
df.head(5)
```

```python
import numpy as np
import pandas as pd

def calculate_risk_score(row):
    score = 0

    # 1. Sleep Patterns (menggunakan huruf kecil untuk pembanding)
    sleep = str(row.get('Sleep Patterns', '')).strip().lower()
    if sleep == 'poor':
        score += 2

    # 2. Physical Activity Level
    activity = str(row.get('Physical Activity Level', '')).strip().lower()
    if activity == 'sedentary':
        score += 2

    # 3. Dietary Habits
    diet = str(row.get('Dietary Habits', '')).strip().lower()
    if diet == 'unhealthy':
        score += 1

    # 4. Smoking Status
    smoking = str(row.get('Smoking Status', '')).strip().lower()
    if smoking == 'current':
        score += 1

    return score

# Hitung ulang skor risiko
df['risk_score'] = df.apply(calculate_risk_score, axis=1)

# Tentukan ulang status risiko
def assign_risk_status(score):
    if score <= 1:
        return 'Low Risk'
    elif score <= 3:
        return 'Moderate Risk'
    else:
        return 'High Risk'

df['risk_status'] = df['risk_score'].apply(assign_risk_status)

# Tampilkan distribusi status risiko yang baru
print(df['risk_status'].value_counts())
```
